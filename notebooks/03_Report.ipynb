# Project Report – Histopathologic Cancer Detection

## Problem Description and Data Overview

This mini-project addresses the challenge presented in the Histopathologic Cancer Detection competition on Kaggle, where the objective is to build a binary image classifier capable of detecting metastatic cancer in small histopathologic image patches. Each image is a 96x96 RGB `.tif` file, and the training dataset consists of over 220,000 labeled examples. The labels are binary: `1` indicates cancerous tissue and `0` indicates the absence of cancer. Alongside the training images is a CSV file that maps each image ID to its label. The test dataset contains unlabeled images for which predictions must be submitted for leaderboard evaluation.

## Exploratory Data Analysis (EDA)

The first step was to perform a structured exploratory analysis of the dataset. First the the image dimensions were verified and and random image samples were reviewed to understand the visual characteristics of cancerous and non-cancerous patches. A class distribution plot revealed a slight imbalance favoring non-cancerous images. Next was also confirmed the integrity of the label file and inspected for missing or duplicate entries. Based on these findings, all images were resized to 64x64 pixels to reduce training time and memory usage, particularly important in early testing when GPU resources were unavailable. These steps helped inform a streamlined plan for preprocessing and model design.

## Model Architecture and Rationale

The chosen model was a lightweight convolutional neural network implemented in PyTorch. It featured two convolutional layers with ReLU activations and max pooling, followed by a flattening layer and two fully connected layers leading to a softmax output. This architecture was intentionally kept simple to accommodate early training on CPU-based environments (all of this occurred because I was totally unaware KAggle provides GPU time to users). Once validated, the model was scaled to a GPU-enabled notebook on Kaggle and trained the same architecture on the full dataset. 

In addition to this main model, variants were tested by increasing the image resolution and expanding the number of filters per layer. This mostlyfound diminishing returns when increasing image size without more complex architectures, ataset size over model depth was prioritized. Hyperparameters such as learning rate and batch size were tuned during training, and stratified sampling were also emplyed to maintain class balance across splits.

## Results and Analysis

The model performed well on both the subset and full training data. When trained on a small sample, validation accuracy reached approximately 83%, which helped confirm the pipeline was functioning correctly. After migrating to a full dataset and GPU environment, validation accuracy improved steadily with each epoch and stabilized above 88% by the end of training. This performance was consistent with visual inspection of the learning curves. 

It was observed that increased training size was more beneficial than deeper architectures. Larger input dimensions and longer training times did not yield proportional accuracy gains confirming that a well-tuned lightweight model can still perform competitively. Notably, overfitting was minimized through appropriate validation strategies and early experimentation with regularization.

## Conclusion

In this project a CNN-based model was successfully built for binary classification of cancerous tissue using histopathologic images.  Although it did not place highly on the Kaggle leaderboard, the model achieved a respectable validation accuracy and produced a valid submission. 

Key learnings included the benefit of full-data training, the trade-offs between input size and training speed and the importance of adapting to hardware constraints. With more time and additional fine-tuning this model could be pushed to be even more accuarte, and perform more efficiently.
